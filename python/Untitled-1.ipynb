```python
import pandas as pd

columns_to_keep = [
    "Date",
    "Time",
    "HomeTeam",
    "AwayTeam",

    # Full-time result
    "FTHG",
    "FTAG",
    "FTR",

    # Half-time result
    "HTHG",
    "HTAG",
    "HTR",

    # Match statistics
    "HS",
    "AS",
    "HST",
    "AST",
    "HF",
    "AF",
    "HC",
    "AC",
    "HY",
    "AY",
    "HR",
    "AR",

    # Betting odds
    "B365H",
    "B365D",
    "B365A",
    "PSH",
    "PSD",
    "PSA"
]

column_mapping = {
    "Date": "MatchDate",
    "Time": "KickoffTime",

    "HomeTeam": "HomeTeam",
    "AwayTeam": "AwayTeam",

    "FTHG": "HomeGoals",
    "FTAG": "AwayGoals",
    "FTR": "Result",

    "HTHG": "HomeHalfTimeGoals",
    "HTAG": "AwayHalfTimeGoals",
    "HTR": "HalfTimeResult",

    "HS": "HomeShots",
    "AS": "AwayShots",
    "HST": "HomeShotsOnTarget",
    "AST": "AwayShotsOnTarget",
    "HF": "HomeFouls",
    "AF": "AwayFouls",
    "HC": "HomeCorners",
    "AC": "AwayCorners",
    "HY": "HomeYellowCards",
    "AY": "AwayYellowCards",
    "HR": "HomeRedCards",
    "AR": "AwayRedCards",

    "B365H": "Bet365HomeOdds",
    "B365D": "Bet365DrawOdds",
    "B365A": "Bet365AwayOdds",

    "PSH": "PinnacleHomeOdds",
    "PSD": "PinnacleDrawOdds",
    "PSA": "PinnacleAwayOdds"
}
```

```python
def load_matches(file_path, league, season):

    df = pd.read_csv(file_path)

    # Keep only the columns needed for our project
    df = df[columns_to_keep].copy()

    # Rename columns
    df = df.rename(columns=column_mapping)

    # Add metadata
    df["League"] = league
    df["Season"] = season

    # Convert date
    df["MatchDate"] = pd.to_datetime(
        df["MatchDate"],
        dayfirst=True
    )

    return df
```

```python
epl_20_21 = load_matches(
    "data/raw/EPL_2020_21.csv",
    "EPL",
    "2020-21"
)

epl_21_22 = load_matches(
    "data/raw/EPL_2021_22.csv",
    "EPL",
    "2021-22"
)

epl_22_23 = load_matches(
    "data/raw/EPL_2022_23.csv",
    "EPL",
    "2022-23"
)

epl_23_24 = load_matches(
    "data/raw/EPL_2023_24.csv",
    "EPL",
    "2023-24"
)

epl_24_25 = load_matches(
    "data/raw/EPL_2024_25.csv",
    "EPL",
    "2024-25"
)

epl_25_26 = load_matches(
    "data/raw/EPL_2025_26.csv",
    "EPL",
    "2025-26"
)
```

```python
laliga_20_21 = load_matches(
    "data/raw/LaLiga_2020_21.csv",
    "La Liga",
    "2020-21"
)

laliga_21_22 = load_matches(
    "data/raw/LaLiga_2021_22.csv",
    "La Liga",
    "2021-22"
)

laliga_22_23 = load_matches(
    "data/raw/LaLiga_2022_23.csv",
    "La Liga",
    "2022-23"
)

laliga_23_24 = load_matches(
    "data/raw/LaLiga_2023_24.csv",
    "La Liga",
    "2023-24"
)

laliga_24_25 = load_matches(
    "data/raw/LaLiga_2024_25.csv",
    "La Liga",
    "2024-25"
)

laliga_25_26 = load_matches(
    "data/raw/LaLiga_2025_26.csv",
    "La Liga",
    "2025-26"
)
```

```python
matches = pd.concat([
    epl_20_21,
    epl_21_22,
    epl_22_23,
    epl_23_24,
    epl_24_25,
    epl_25_26,
    laliga_20_21,
    laliga_21_22,
    laliga_22_23,
    laliga_23_24,
    laliga_24_25,
    laliga_25_26
], ignore_index=True)
```

```python
print("Total matches:", len(matches))

print("\nMatches by league:")
print(matches["League"].value_counts())

print("\nMatches by season:")
print(matches["Season"].value_counts())

print("\nMatches by league and season:")
print(
    matches.groupby(["League", "Season"])
    .size()
)
```

```python
home = matches[[
    "League",
    "Season",
    "MatchDate",
    "HomeTeam",
    "AwayTeam",

    # Full-time
    "HomeGoals",
    "AwayGoals",

    # Half-time
    "HomeHalfTimeGoals",
    "AwayHalfTimeGoals",

    # Match statistics
    "HomeShots",
    "HomeShotsOnTarget",
    "HomeCorners",
    "HomeFouls",
    "HomeYellowCards",
    "HomeRedCards",

    # Betting odds
    "Bet365HomeOdds",
    "Bet365AwayOdds",
    "PinnacleHomeOdds",
    "PinnacleAwayOdds"
]].copy()

home = home.rename(columns={
    "HomeTeam": "Team",
    "AwayTeam": "Opponent",

    "HomeGoals": "GoalsScored",
    "AwayGoals": "GoalsConceded",

    "HomeHalfTimeGoals": "HalfTimeGoalsScored",
    "AwayHalfTimeGoals": "HalfTimeGoalsConceded",

    "HomeShots": "Shots",
    "HomeShotsOnTarget": "ShotsOnTarget",
    "HomeCorners": "Corners",
    "HomeFouls": "Fouls",
    "HomeYellowCards": "YellowCards",
    "HomeRedCards": "RedCards",

    "Bet365HomeOdds": "Bet365Odds",
    "Bet365AwayOdds": "OpponentBet365Odds",

    "PinnacleHomeOdds": "PinnacleOdds",
    "PinnacleAwayOdds": "OpponentPinnacleOdds"
})

home["Venue"] = "Home"
```

```python
away = matches[[
    "League",
    "Season",
    "MatchDate",
    "AwayTeam",
    "HomeTeam",

    # Full-time
    "AwayGoals",
    "HomeGoals",

    # Half-time
    "AwayHalfTimeGoals",
    "HomeHalfTimeGoals",

    # Match statistics
    "AwayShots",
    "AwayShotsOnTarget",
    "AwayCorners",
    "AwayFouls",
    "AwayYellowCards",
    "AwayRedCards",

    # Betting odds
    "Bet365AwayOdds",
    "Bet365HomeOdds",
    "PinnacleAwayOdds",
    "PinnacleHomeOdds"
]].copy()

away = away.rename(columns={
    "AwayTeam": "Team",
    "HomeTeam": "Opponent",

    "AwayGoals": "GoalsScored",
    "HomeGoals": "GoalsConceded",

    "AwayHalfTimeGoals": "HalfTimeGoalsScored",
    "HomeHalfTimeGoals": "HalfTimeGoalsConceded",

    "AwayShots": "Shots",
    "AwayShotsOnTarget": "ShotsOnTarget",
    "AwayCorners": "Corners",
    "AwayFouls": "Fouls",
    "AwayYellowCards": "YellowCards",
    "AwayRedCards": "RedCards",

    "Bet365AwayOdds": "Bet365Odds",
    "Bet365HomeOdds": "OpponentBet365Odds",

    "PinnacleAwayOdds": "PinnacleOdds",
    "PinnacleHomeOdds": "OpponentPinnacleOdds"
})

away["Venue"] = "Away"
```

```python
team_matches = pd.concat(
    [home, away],
    ignore_index=True
)

print("Team-match records:", len(team_matches))
```

```python
team_matches["GoalDifference"] = (
    team_matches["GoalsScored"]
    - team_matches["GoalsConceded"]
)

team_matches["Result"] = "Draw"

team_matches.loc[
    team_matches["GoalsScored"] > team_matches["GoalsConceded"],
    "Result"
] = "Win"

team_matches.loc[
    team_matches["GoalsScored"] < team_matches["GoalsConceded"],
    "Result"
] = "Loss"
```

```python
team_matches["Points"] = 1

team_matches.loc[
    team_matches["Result"] == "Win",
    "Points"
] = 3

team_matches.loc[
    team_matches["Result"] == "Loss",
    "Points"
] = 0
```

```python
team_matches["ShotConversion"] = (
    team_matches["GoalsScored"]
    / team_matches["Shots"].replace(0, pd.NA)
)

team_matches["OnTargetConversion"] = (
    team_matches["GoalsScored"]
    / team_matches["ShotsOnTarget"].replace(0, pd.NA)
)
```

```python
team_matches["HalfTimeResult"] = "Draw"

team_matches.loc[
    team_matches["HalfTimeGoalsScored"] > team_matches["HalfTimeGoalsConceded"],
    "HalfTimeResult"
] = "Win"

team_matches.loc[
    team_matches["HalfTimeGoalsScored"] < team_matches["HalfTimeGoalsConceded"],
    "HalfTimeResult"
] = "Loss"
```

```python
team_matches["IsUnderdog"] = (
    team_matches["Bet365Odds"]
    > team_matches["OpponentBet365Odds"]
)
```

```python
team_matches["UnderdogWon"] = (
    (team_matches["IsUnderdog"] == True)
    & (team_matches["Result"] == "Win")
)
```

```python
team_matches["PinnacleUnderdog"] = (
    team_matches["PinnacleOdds"]
    > team_matches["OpponentPinnacleOdds"]
)
```
